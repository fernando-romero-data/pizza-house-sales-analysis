# 18.7 Build Customer Density Map

## Objective

Create the customer geographic distribution map.

This section visualizes the geographic footprint of the Pizza House customer database using the latitude and longitude coordinates generated in Workbook 17. The map highlights where customers are concentrated throughout the service area and establishes the foundation for neighborhood-level marketing analysis.

The Customer Density Map provides an executive view of customer distribution before drilling into ZIP code and city-level summaries.

Tableau Steps

Worksheet:
02_Customer_Density_Map

1. Double-click Latitude.

2. Double-click Longitude.

3. Drag customer_id_clean to Detail on the Marks card.

4. Change the Marks type to:
   Circle

5. Adjust the circles:

   • Size: Small–Medium
   • Opacity: 40–50%

6. Drag the following fields to Tooltip:

   • address_raw
   • city
   • zip_code
   • activation_status
   • campaign_touch_count

7. Fit the map to:
   Entire View

Expected Result

• Approximately 9,500+ customer locations are displayed.
• Customer clusters become visible throughout the Sacramento service area.
• Areas with higher customer density naturally appear darker due to overlapping points.

# 18.8 Build Customers by ZIP Code

## Objective

Visualize the customer distribution by ZIP code.

This section summarizes the Pizza House customer database by ZIP code to identify where customers are geographically concentrated. The visualization highlights the neighborhoods with the largest customer populations and provides a foundation for future direct mail campaigns, localized promotions, and territory planning.

The ZIP code summary complements the customer density map by presenting an aggregated geographic view that is easier to compare across service areas.

Tableau Steps

Worksheet:
03_Customers_by_ZIP

1. Drag zip_code to Rows.

2. Drag customer_id_clean to Columns.

3. Change customer_id_clean to:
   Count (Distinct)

4. Click the Sort Descending button.

5. Show Mark Labels.

6. Rename the worksheet title:
   Customers by ZIP Code

7. Edit the x-axis title:
   Distinct Customers

8. Fit the view:
   Entire View

9. Format the chart:
   • Marks: Bar
   • Keep Tableau default blue
   • Remove grid lines (optional)
   • Increase font size if needed for readability

10. Add the following fields to Tooltip:
    • zip_code
    • customer_id_clean (Count Distinct)

Expected Result

• Horizontal bar chart sorted from highest to lowest customer count.
• 29 ZIP codes represented.
• ZIP Code 95820 appears as the largest customer segment.
• The chart clearly highlights the primary service areas for future marketing efforts.

   Verified Response Rate by ZIP Code

   Verified Response Rate

    Entire View

• ZIP Code
