## 1.7 AOV Optimization Strategy — Tactical Implementation

- Objective:
  - Translate AOV insights into actionable tactics
  - Increase revenue per order without adding operational strain

- Core Principle:
  - Drive higher spend per transaction while maintaining speed and margin

## 1.7.1 Strategic Framework

- Key Insight:
  - Current AOV ≈ $30.70
  - Small increases in AOV produce significant revenue impact

- Approach:
  - Use structured incentives instead of blanket discounts
  - Align pricing tactics with operational constraints

## 1.7.2 Spend Threshold Incentives (Menu-Aligned)

- Strategy:
  - Use realistic price jumps based on pizza sizing and add-ons

- Example (Based on Menu Behavior):

  - Spend 25 → free garlic bread  
    (Typical: Medium Pizza ≈ 22–25 → encourages add-on)

  - Spend 35 → free 2-liter soda  
    (Typical: Large Pizza + side → pushes into full meal)

  - Spend 45 → free wings or bundle upgrade  
    (Typical: Family-size orders)

- Why It Works:
  - Aligns with natural ordering tiers:
    - Solo → Medium
    - Couple → Large + side
    - Family → multi-item order

- Impact:
  - Converts borderline orders into higher AOV tiers

## 1.7.3 Bundle Optimization (Menu-Specific)

- Strategy:
  - Build bundles around your highest-frequency items

- Example Bundles:

  - Family Deal:
    - 1 Large Specialty Pizza + Garlic Bread + 2-Liter → 31–35 range

  - Double Pizza Deal:
    - 2 Medium Pizzas → 32–36 range

  - Party Bundle:
    - 2 Large Pizzas + Wings + 2-Liter → 50–55 range

- Why It Works:
  - Mirrors real ordering behavior (groups, families)
  - Simplifies decision-making
  - Increases total ticket size

- Operational Benefit:
  - Predictable orders = faster kitchen execution

## 1.7.4 Default Upsells (Menu-Driven)

- Strategy:
  - Focus upsells on high-margin, easy-add items

- Best Upsell Targets:

  - Garlic Bread:
    - “Add garlic bread for 5?”

  - Size Upgrade:
    - “Upgrade to large for 3 more?”

  - Wings:
    - “Add wings for 8–10?”

  - Drinks:
    - “Add a 2-liter for 3–4?”

- Why It Works:
  - Low friction additions
  - Minimal impact on kitchen complexity
  - High contribution margin items

- Expected Impact:
  - 5–10% AOV increase through consistent execution

## 1.7.5 Menu Anchoring (Pricing Psychology)

- Strategy:
  - Structure pricing tiers to encourage natural upgrades

- Example (Menu Behavior):

  - Medium Pizza → ~22–25  
  - Large Pizza → ~26–29  

- Insight:
  - Small price gaps (3–4 difference) drive customers to choose larger sizes

- Why It Works:
  - Customers perceive higher value in upgrading
  - Increases AOV without additional items or complexity

- Impact:
  - Shifts order mix toward higher-value items

## 1.7.6 Strategic Takeaway

- AOV growth should be driven by:
  - Spend thresholds
  - Bundles
  - Upsells
  - Pricing structure

- Avoid:
  - Blanket discounts
  - Free items without conditions

- Key Principle:
  - Increase revenue per order without increasing operational strain

- Result:
  - Higher AOV
  - Improved margins
  - Faster, more efficient ordering flow

## 2.1 AOV Expansion — Revenue Impact Estimation

- Objective:
  - Estimate revenue lift from increasing Average Order Value (AOV)
  - Quantify the impact of bundles, upsells, and spend-threshold incentives

## 2.1.0 Load Order Data

- Using pre-concatenated full-year dataset
- Required columns:
  - order_date
  - order_id
  - order_total

- Ensure:
  - Column names are standardized
  - Order Total is numeric
  - No missing core fields

```python
import pandas as pd

# Load dataset
df = pd.read_csv("../data/cleaned/orders_2025_full.csv")

# Standardize column names
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")

# Keep required columns for Section 2
df = df[["order_date", "order_id", "order_total"]].copy()

# Ensure numeric order total
df["order_total"] = pd.to_numeric(df["order_total"], errors="coerce")

# Drop rows missing core fields
df = df.dropna(subset=["order_date", "order_id", "order_total"])

df.head()
```

## 2.1.0a Create Datetime & Time Features

- Convert raw timestamp into localized datetime
- Standardize timezone to America/Los_Angeles
- Create reusable time-based fields for downstream analysis:
  - hour
  - day of week
  - weekend flag
  - peak / off-peak flags

- This supports later sections without rebuilding the pipeline

```python
# Create localized datetime
df["order_datetime"] = pd.to_datetime(df["order_date"], utc=True)
df["order_datetime"] = df["order_datetime"].dt.tz_convert("America/Los_Angeles")

# Create reusable time features
df["hour"] = df["order_datetime"].dt.hour
df["order_dow"] = df["order_datetime"].dt.day_name()
df["is_weekend"] = df["order_dow"].isin(["Saturday", "Sunday"])

# Create reusable business flags
df["peak_flag"] = (df["hour"] >= 16) & (df["hour"] < 19)
df["off_peak_flag"] = df["hour"] < 16

df[["order_datetime", "hour", "order_dow", "is_weekend", "peak_flag", "off_peak_flag"]].head()
```

## 2.1.0b Create Core Baseline Metrics

- Compute reusable baseline metrics for Section 2:
  - Average Order Value (AOV)
  - Total Orders

- These metrics will be reused across all revenue scenario calculations

```python
baseline_aov = df["order_total"].mean()
total_orders = df["order_id"].nunique()

baseline_aov, total_orders
```

## 2.1.1 Baseline Metrics

- Average Order Value (AOV): $30.70
- Total Orders (Annual): 50,487

- These metrics form the foundation for AOV scenario modeling

```python
baseline_aov = round(baseline_aov, 2)
total_orders = int(total_orders)

baseline_aov, total_orders
```

## 2.1.2 AOV Expansion Scenarios

- Scenario Design:
  - Conservative Case: +5% AOV
  - Moderate Case: +10% AOV
  - Aggressive Case: +15% AOV

- Key Insight:
  - Even small increases in AOV generate meaningful revenue impact
  - No additional labor or capacity is required
  - This is a high-margin growth lever

```python
aov_scenarios = {
    "5% Increase": 0.05,
    "10% Increase": 0.10,
    "15% Increase": 0.15
}

aov_results = []

for label, pct in aov_scenarios.items():
    new_aov = baseline_aov * (1 + pct)
    revenue_lift = (new_aov - baseline_aov) * total_orders

    aov_results.append({
        "Scenario": label,
        "New AOV ($)": round(new_aov, 2),
        "Revenue Lift ($)": round(revenue_lift, 2)
    })

aov_results_df = pd.DataFrame(aov_results)

aov_results_df
```

## 2.1.3 Interpretation

- Key Findings:
  - AOV growth creates substantial top-line revenue lift
  - This is the most operationally efficient revenue lever in the workbook

- Business Insight:
  - Upsells, bundles, and spend-threshold incentives can increase revenue without requiring more orders
  - Compared with staffing-based levers, this is lower complexity and higher margin

- Strategic Takeaway:
  - AOV expansion should be prioritized as a first-line revenue strategy

## 2.2 Peak Throughput — Revenue Impact Estimation

- Objective:
  - Quantify revenue gains from improving throughput during peak hours (4pm–7pm)
  - Estimate how many additional orders can be fulfilled with better staffing and kitchen efficiency

## 2.2.0 Peak Hour Demand Context

- Key Insight:
  - Peak demand occurs between 4pm–7pm
  - Kitchen capacity limits total orders fulfilled
  - Revenue is constrained by throughput, not demand

- Assumption:
  - Additional staffing and better kitchen execution increase fulfilled orders per hour

## 2.2.1 Baseline Peak Metrics

- Measure current order volume during peak hours
- Use full-year local-time order data
- Establish baseline revenue per peak hour

```python
# Filter peak-hour orders
peak_df = df[df["peak_flag"]].copy()

# Baseline peak metrics
peak_orders_per_hour = peak_df.groupby("hour")["order_id"].nunique().mean()
peak_revenue_per_hour = peak_orders_per_hour * baseline_aov

peak_orders_per_hour, peak_revenue_per_hour
```

## 2.2.2 Throughput Improvement Scenarios

- Scenario Design:
  - Conservative: +2 additional orders per hour
  - Moderate: +5 additional orders per hour
  - Aggressive: +8 additional orders per hour

- Assumption:
  - Improved staffing and kitchen flow increase fulfilled orders during the 4pm–7pm window

- Modeling Note:
  - Peak window assumed to last 3 hours per day

```python
throughput_scenarios = {
    "Conservative (+2 orders/hr)": 2,
    "Moderate (+5 orders/hr)": 5,
    "Aggressive (+8 orders/hr)": 8
}

throughput_results = []
peak_hours_per_day = 3

for label, extra_orders in throughput_scenarios.items():
    incremental_revenue_per_hour = extra_orders * baseline_aov
    daily_revenue_lift = incremental_revenue_per_hour * peak_hours_per_day
    annual_revenue_lift = daily_revenue_lift * 365

    throughput_results.append({
        "Scenario": label,
        "Extra Orders/Hour": extra_orders,
        "Revenue Lift / Hour ($)": round(incremental_revenue_per_hour, 2),
        "Revenue Lift / Day ($)": round(daily_revenue_lift, 2),
        "Revenue Lift / Year ($)": round(annual_revenue_lift, 2)
    })

throughput_df = pd.DataFrame(throughput_results)

throughput_df
```

## 2.2.3 Interpretation

- Key Findings:
  - Small increases in peak-hour throughput create meaningful revenue upside
  - Peak hours represent the highest-return operating window for staffing improvements

- Business Insight:
  - This lever depends on operational execution, not pricing
  - Revenue upside is repeatable because peak demand is consistent and predictable

- Strategic Takeaway:
  - Improving kitchen throughput during 4pm–7pm is one of the clearest revenue unlocks in the business

- Use existing off-peak flag created earlier in the pipeline
- Establish baseline for demand-expansion scenarios

```python
# Filter off-peak orders
off_peak_orders = df[df["off_peak_flag"]].copy()

# Baseline off-peak metrics
off_peak_orders_per_hour = off_peak_orders.groupby("hour")["order_id"].nunique().mean()
off_peak_revenue_per_hour = off_peak_orders_per_hour * baseline_aov

off_peak_orders_per_hour, off_peak_revenue_per_hour
```

## 2.3.2 Off-Peak Demand Scenarios

- Scenario Design:
  - Conservative: +2 additional orders per hour
  - Moderate: +4 additional orders per hour
  - Aggressive: +6 additional orders per hour

- Demand Drivers:

  - Lunch specials
  - Time-based promotions

```python
off_peak_scenarios = {
    "Conservative (+2 orders/hr)": 2,
    "Moderate (+4 orders/hr)": 4,
    "Aggressive (+6 orders/hr)": 6
}

off_peak_results = []
off_peak_hours_per_day = 8

for label, extra_orders in off_peak_scenarios.items():
    incremental_revenue_per_hour = extra_orders * baseline_aov
    daily_revenue_lift = incremental_revenue_per_hour * off_peak_hours_per_day
    annual_revenue_lift = daily_revenue_lift * 365

    off_peak_results.append({
        "Scenario": label,
        "Extra Orders/Hour": extra_orders,
        "Revenue Lift / Hour ($)": round(incremental_revenue_per_hour, 2),
        "Revenue Lift / Day ($)": round(daily_revenue_lift, 2),
        "Revenue Lift / Year ($)": round(annual_revenue_lift, 2)
    })

off_peak_results_df = pd.DataFrame(off_peak_results)

off_peak_results_df
```

## 2.3.3 Interpretation

- Key Findings:
